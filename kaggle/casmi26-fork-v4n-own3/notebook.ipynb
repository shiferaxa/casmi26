## Why each change helps, and where it hurts

**PubChem join (new).** The main engine ranks candidates from a pool of about 710,000 structures (training structures
and COCONUT). When the answer is not in that pool, it can only come from the PubChem-only channel, and in the reference
those proposals are written into fixed slots of the finished list (4, 8, 12, ...). So an answer that exists only in
PubChem can never be first, and its place depends on a single fingerprint score.

The join hands the best PubChem-only proposals to the main engine as ordinary candidates. They get the same features
and the same ranker as pool candidates, and the forward models score them like the rest of the list. Molecules with a
good library match are skipped, so their lists stay as they were.
- Good: an outside answer can be ranked first. It is the first change of ours that moved the leaderboard by more than
  a few thousandths (0.409 to 0.420).
- Good: nothing inside the public engine is edited, and a failure on the scoring run falls back to the list without
  the join for that molecule.
- Bad: most joined candidates are wrong, so they must not displace correct pool answers; that is what the library gate
  is for.
- Bad: more candidates per molecule cost run time and memory (the scoring run takes 3 to 4 hours on a T4), and a
  structure that is in neither the pool nor PubChem is still out of reach.

**Popularity prior.** Most errors of this pipeline are between isomers: candidates with the same formula, where the
spectrum evidence is often too weak to decide. Among such candidates the compound with more PubChem substance records
and more PubMed mentions is more often the one somebody actually measured. The prior adds
`mu * (log1p(substances) + log1p(PubMed))` to the z-scored ranker score, for pool candidates only (generated
candidates have no record and keep their slots).
- Good: clear gain for well-studied natural products; costs nothing at run time.
- Bad: it is not spectrum evidence. For ordinary or synthetic compounds it pushes the famous isomer above the right
  one; on a synthetic library it was worse than no prior in our local checks.

**Library gate for the forward models.** ICEBERG and GLACIER predict a spectrum for each candidate and re-order
same-formula candidates by similarity to the measured one. When the library already holds a near-identical reference
spectrum (`lib_max >= 0.9`) the measured match is stronger evidence than a simulated spectrum. In a local check the
forward models moved the correct structure down for 18 of 400 exact library hits.
- Good: removes a failure mode on the molecules that should be the easiest.
- Bad: no measurable leaderboard change on its own (0.404 before and after). It is a safety change, not a gain.

**Fragment re-score in the gap.** ICEBERG and GLACIER only score [M+H]+ and [M+Na]+ spectra. Molecules measured only
as other adducts or in negative mode get no isomer evidence at all. For those molecules only, a MetFrag-style score
(how much of the peak intensity can be explained by breaking one or two bonds of the candidate) is added.
- Good: gives some evidence where there was none.
- Bad: it is a weak signal. Applied to all molecules it dilutes the forward models and scored **0.390**, clearly worse.

## What did not work
- Fragment re-score for every molecule: 0.390.
- Popularity prior inside the PubChem-only channel, using the public popularity arrays: 0.393 to 0.405.
- Re-ordering pool candidates in other ways (different tie handling in the ranker, a gate on the engine fusion,
  GLACIER on [M+H]+ spectra only): 0.405 to 0.408, no better than 0.409.

## Known weaknesses
- **Local validation is misleading.** The public models (fingerprint network, feature models, ranker) were trained on
  all of `train.parquet`, so any hold-out drawn from train is recognised: the ranker puts the truth first about 93% of
  the time locally. We use local numbers only to check that a change does no harm.
- **The public leaderboard is small.** Differences of a few thousandths are inside the noise; the steps in the table
  above are single submissions.
- **Isomer order is the main loss** when the true structure is in the candidate list.
- **Engine 2** (analog propagation) is strong when the library contains the compound or close analogs and contributes
  little otherwise.

## How it runs
The commit run scores only 12 molecules of the visible test (`SMOKE_N`), which is why it finishes in about 20 minutes;
the visible test is a placeholder drawn from train, so its score means nothing. The scoring rerun on the hidden test
runs everything (3 to 4 hours on a T4). No internet; all inputs are public Kaggle datasets. Nothing is keyed to the
visible test set.

```python
# setup: offline rdkit wheel, input discovery, MANIFEST check, embedded sources written to /kaggle/working/ours
import os, sys, glob, re, subprocess, time, json, hashlib, pickle
T0 = time.time()
WORK = '/kaggle/working'
STAGE = os.path.join(WORK, 'stage_cache')
OURS = os.path.join(WORK, 'ours')
os.makedirs(os.path.join(WORK, 'numba_cache'), exist_ok=True)
os.environ['NUMBA_CACHE_DIR'] = os.path.join(WORK, 'numba_cache')

def find(pattern):
    hits = sorted(glob.glob(f'/kaggle/input/**/{pattern}', recursive=True), key=len)
    if not hits:
        raise FileNotFoundError(pattern)
    return hits[0]

def sha256(path, n=1 << 22):
    h = hashlib.sha256()
    with open(path, 'rb') as f:
        for b in iter(lambda: f.read(n), b''):
            h.update(b)
    return h.hexdigest()

def dump(name, obj):
    if CFG['DUMP_STAGES']:
        os.makedirs(STAGE, exist_ok=True)
        json.dump(obj, open(os.path.join(STAGE, f'{name}.json'), 'w'))

def log(*a):
    print(f'[{time.time()-T0:6.0f}s]', *a, flush=True)

tag = f'cp{sys.version_info.major}{sys.version_info.minor}'
whl = [w for w in glob.glob('/kaggle/input/**/rdkit-*.whl', recursive=True) if tag in w]
log('rdkit wheel:', whl)
r = subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '--no-index', '--no-deps', whl[0]], capture_output=True, text=True)
print(r.stdout[-500:], r.stderr[-500:])
import rdkit; log('rdkit', rdkit.__version__)
assert rdkit.__version__ == '2026.03.3', 'metric is pinned to RDKit 2026.03.3'

V4CODE = os.path.dirname(os.path.dirname(find('casmi26-v4b-models/**/casmi/engine.py')))
V4MOD = os.path.dirname(find('casmi26-v4b-models/**/MANIFEST.json'))
V3CODE = os.path.dirname(os.path.dirname(find('casmi26-v3-models/**/casmi/engine.py')))
V3MOD = os.path.dirname(find('casmi26-v3-models/**/ranker_0.pkl'))
POOL_DIR = os.path.dirname(find('pool_meta.parquet'))
PC_DIR = os.path.dirname(find('pc_smiles.npy'))
COMP = os.path.dirname(find('test.parquet'))
MAN = json.load(open(os.path.join(V4MOD, 'MANIFEST.json')))
bad = []
for rel, h in sorted(MAN['files'].items()):
    p = os.path.join(V4CODE, rel[len('code/'):]) if rel.startswith('code/') else os.path.join(V4MOD, rel)
    if not os.path.exists(p) or sha256(p) != h:
        bad.append(rel)
assert not bad, f'dataset files differ from MANIFEST.json: {bad}'
log('V4CODE', V4CODE, '| V4MOD', V4MOD, '| V3CODE', V3CODE, '| POOL', POOL_DIR, '| PC', PC_DIR, '| COMP', COMP)
log(f"{len(MAN['files'])} files match MANIFEST.json; families {MAN['families']}")

for _rel, _src in EMBED.items():        # EMBED is injected by tools/build_notebook.py from nb/embed/
    _p = os.path.join(OURS, _rel); os.makedirs(os.path.dirname(_p), exist_ok=True)
    open(_p, 'w').write(_src)
sys.path.insert(0, OURS)
import fusion_core
log('embedded sources:', sorted(EMBED))
json.dump(CFG, open(os.path.join(WORK, 'cfg.json'), 'w'), indent=1)
```

## 4. Commit run or scoring rerun
The visible `test.parquet` is a placeholder. On a commit run only `SMOKE_N` molecules are processed, to prove that the
pipeline runs end to end; on the scoring rerun (detected by the environment variable and by the test file's signature)
everything runs. A validation mode exists as well, with the caveat from the header: it is leaked.


```python
# SMOKE switch: the commit run (visible test = train examples) scores SMOKE_N molecules unless FULL_ON_COMMIT;
# the scoring rerun (hidden test) always runs everything. VALIDATION (commit runs only) swaps in a held-out test.
import pandas as pd
_te = pd.read_parquet(os.path.join(COMP, 'test.parquet'), columns=['molecule_id', 'spectrum_id'])
_sig = hashlib.md5(','.join(sorted(_te.spectrum_id.astype(str))).encode()).hexdigest()
IS_RERUN = bool(os.getenv('KAGGLE_IS_COMPETITION_RERUN')) or _sig != '654e030e4173780e945252b4a5adf70b'
ICE_BUDGET = CFG['ICE_BUDGET']

TEST_ADDUCTS = ['[M+H]+', '[M+NH4]+', '[M-H2O+H]+', '[M-2H2O+H]+', '[M+Na]+', '[M+K]+',
                '[M-H]-', '[M-H2O-H]-', '[M+CH2O2-H]-', '[M+Cl]-']

def fold0_np_keys():
    """Structures that the fold-safe FPNet (dmitriigluzdov/casmi26-fold-safe-fpnet, fold 'hold') never saw and that
    are in a natural-product catalogue (pool_np > 0: COCONUT / LOTUS / NPAtlas). Ordinary natural products: median
    popularity 5.4 against 13.0 for the famous enveda-np-examples."""
    import numpy as np
    sp = pd.read_parquet(find('casmi26-fold-safe-fpnet/**/split.parquet'))
    pm = pd.read_parquet(os.path.join(POOL_DIR, 'pool_meta.parquet'), columns=['key', 'pid'])
    npf = np.load(find('casmi26-pubchem-popularity-prior/**/pool_np.npy'))
    nat = set(pm.key.values[npf[pm.pid.values] > 0])
    h = sp[(sp.fold == 'hold') & sp.key.isin(nat)]
    return set(h.inchikey14)


def build_validation(comp, out, lib, n, seed, max_spec, mode='all', only=None):
    """Hold out n structures: their spectra (test adducts, base peak >= 1000 counts, <= max_spec per molecule) become
    test.parquet, and EVERY train row with those InChIKey14s is dropped from train.parquet, so the library cannot
    answer them (candidate pools still contain them: a class-2-like setting). mode='lib' drops only the held-out
    `lib` rows, so spectra of the same structures from OTHER libraries stay searchable: a class-1-like setting (a new
    timsTOF measurement of a compound with public reference spectra). lib=None with `only` (a set of InChIKey14s)
    draws the structures from that set and their test spectra from any library. Writes labels too."""
    import numpy as np, pyarrow as pa, pyarrow.parquet as pq, pyarrow.compute as pc
    src = os.path.join(comp, 'train.parquet')
    meta = pq.read_table(src, columns=['ingest_lib', 'inchikey14', 'adduct']).to_pandas()
    ok = meta.adduct.isin(TEST_ADDUCTS)
    if lib is not None:
        ok &= meta.ingest_lib == lib
    if only is not None:
        ok &= meta.inchikey14.isin(only)
    keys = sorted(meta.loc[ok, 'inchikey14'].unique())
    # audit 2026-10-02: the metric compares tautomer-canonical keys, so sample one structure per metric key and purge
    # every training InChIKey14 that shares a held-out metric key (train_structs.parquet carries key per structure)
    _st = pd.read_parquet(os.path.join(POOL_DIR, 'train_structs.parquet'), columns=['inchikey14', 'key'])
    ik2key = dict(zip(_st.inchikey14, _st.key))
    order = [keys[i] for i in np.random.RandomState(seed).permutation(len(keys))]
    keys, seen = [], set()
    for k in order:
        mk = ik2key.get(k, k)
        if mk not in seen:
            seen.add(mk); keys.append(k)
        if len(keys) >= n:
            break
    hold = pa.array(keys)
    purge = pa.array(sorted(set(keys) | set(_st.inchikey14[_st.key.isin(seen)])))   # held-out keys and their twins
    os.makedirs(out, exist_ok=True)
    pf, writer, parts = pq.ParquetFile(src), None, []
    for rg in range(pf.num_row_groups):
        t = pf.read_row_group(rg)
        held = pc.is_in(t['inchikey14'], value_set=hold)
        twin = pc.is_in(t['inchikey14'], value_set=purge)
        in_lib = pc.equal(t['ingest_lib'], lib) if lib is not None else held
        drop = twin if mode == 'all' else pc.and_(held, in_lib)
        kept = t.filter(pc.invert(drop))
        writer = writer or pq.ParquetWriter(os.path.join(out, 'train.parquet'), kept.schema)
        writer.write_table(kept)
        q = t.filter(pc.and_(pc.and_(held, in_lib), pc.is_in(t['adduct'], value_set=pa.array(TEST_ADDUCTS))))
        if q.num_rows:
            parts.append(q)
    writer.close()
    q = pa.concat_tables(parts).to_pandas()
    q = q[q.base_peak_intensity.isna() | (q.base_peak_intensity >= 1000)]
    if lib is None:                               # many libraries per structure: a seeded random subset, not the first rows
        q = q.sample(frac=1.0, random_state=seed).sort_values('inchikey14', kind='stable')
    q = q.groupby('inchikey14', group_keys=False).head(max_spec).reset_index(drop=True)
    q['molecule_id'] = ['v_' + hashlib.md5(k.encode()).hexdigest()[:8] for k in q.inchikey14]
    q['spectrum_id'] = [f'vs_{i:06d}' for i in range(len(q))]
    if lib is not None:                           # enveda-np-examples really is timsTOF; other libraries keep their label
        q['instrument_type'] = 'timsTOF'
    test_cols = ['molecule_id', 'spectrum_id', 'ms2_mzs', 'ms2_normalized_intensities', 'base_peak_intensity', 'adduct',
                 'ionization_mode', 'instrument_type', 'precursor_mz', 'collision_energy_ev', 'collision_energy_orig',
                 'collision_energy_orig_units']
    q[test_cols].to_parquet(os.path.join(out, 'test.parquet'), index=False)
    lab = q.groupby('molecule_id').agg(smiles=('normalized_smiles', 'first'), inchikey14=('inchikey14', 'first')).reset_index()
    pd.DataFrame({'molecule_id': lab.molecule_id, 'smiles': 'CCO'}).to_csv(os.path.join(out, 'sample_submission.csv'), index=False)
    os.makedirs(STAGE, exist_ok=True)
    lab.to_csv(os.path.join(STAGE, 'val_labels.csv'), index=False)
    return len(lab), len(q)

if CFG['VALIDATION'] and not IS_RERUN:
    _t0 = time.time()
    if CFG['VAL_SET'] == 'fold0_np':
        _nm, _ns = build_validation(COMP, os.path.join(WORK, 'val_comp'), None, CFG['VAL_N'], CFG['VAL_SEED'],
                                    CFG['VAL_MAX_SPEC'], 'all', only=fold0_np_keys())
    else:
        _nm, _ns = build_validation(COMP, os.path.join(WORK, 'val_comp'), CFG['VAL_LIB'], CFG['VAL_N'], CFG['VAL_SEED'],
                                    CFG['VAL_MAX_SPEC'], CFG['VAL_PURGE'])
    COMP = os.path.join(WORK, 'val_comp')
    log(f'VALIDATION {CFG["VAL_SET"]}: {_nm} held-out molecules / {_ns} spectra, purge={CFG["VAL_PURGE"]} ({time.time()-_t0:.0f}s)')
elif not IS_RERUN and not CFG['FULL_ON_COMMIT']:
    SM = os.path.join(WORK, 'smoke_comp'); os.makedirs(SM, exist_ok=True)
    _full = pd.read_parquet(os.path.join(COMP, 'test.parquet'))
    _keep = sorted(_full.molecule_id.unique())[:CFG['SMOKE_N']]
    _full[_full.molecule_id.isin(_keep)].to_parquet(os.path.join(SM, 'test.parquet'))
    for _f in ['train.parquet', 'sample_submission.csv']:
        if not os.path.exists(os.path.join(SM, _f)):
            os.symlink(os.path.join(COMP, _f), os.path.join(SM, _f))
    COMP = SM
    ICE_BUDGET = CFG['SMOKE_ICE_BUDGET']
log('RERUN', IS_RERUN, '| sig', _sig, '| COMP', COMP, '| ICE_BUDGET', ICE_BUDGET, '| FULL_ON_COMMIT', CFG['FULL_ON_COMMIT'],
    '| VALIDATION', CFG['VALIDATION'])
```

## 8. Base lists
For every molecule: candidates from the pool (plus generated ones and, when the molecule has no good library match,
the joined PubChem candidates), ranked by the LightGBM ranker in one list. The top `TOPN` are kept with their scores,
formulas and `lib_max`, the best similarity to a library spectrum. If the join fails for a molecule on the scoring
run, that molecule is processed without it.


```python
# all molecules -> base lists: BASE[mid] = [smiles, keys, lib_max, ranker scores, formulas, pool ids] (top TOPN, key-deduped;
# pool id < 0 = generated candidate)
te = pd.read_parquet(os.path.join(COMP, 'test.parquet'))
te['nm'] = chem.neutral_mass(te.precursor_mz.values.astype(np.float64), te.adduct.values)
mols = list(te.groupby('molecule_id', sort=False))
BASE, n_err, n_rk_err, n_rk_diff, n_join, n_join_mol, n_join_err = {}, 0, 0, 0, 0, 0, 0
for gi, (mid, sub) in enumerate(mols):
    nms = sub.nm.values[np.isfinite(sub.nm.values)]
    smis, keys, lib_max, scs, forms, pids = [], [], 0.0, [], [], []
    if len(nms):
        target = float(np.median(nms))
        spectra = []
        for r in sub.itertuples():
            ce = r.collision_energy_ev
            ce_arr = np.atleast_1d(np.asarray(ce, dtype=float)) if ce is not None else np.zeros(0)
            spectra.append(dict(mz=np.asarray(r.ms2_mzs, np.float64), it=np.asarray(r.ms2_normalized_intensities, np.float64),
                                mode=1 if r.ionization_mode == 'positive' else -1, adduct=r.adduct,
                                prec=float(r.precursor_mz), ce=float(ce_arr.mean()) if len(ce_arr) else np.nan,
                                ce_n=int(len(ce_arr)) if len(ce_arr) else 1))
        try:
            recs, inj, _out = [], None, {}
            if CFG['PC_JOIN_N'] > 0:
                try:
                    _p = PC_DEEP.get(mid) or PC_DEEP.get(str(mid))
                    # smoke commit runs join every molecule (the smoke molecules are library hits): a bug must fail the commit
                    _gate = CFG['PC_JOIN_MAX_LIB'] if (IS_RERUN or CFG['VALIDATION']) else None
                    if _p and (_gate is None or pc_join.library_max(E, spectra, target) < _gate):
                        _ix = pc_join.select_injection(_p, CFG['PC_JOIN_N'], min_rel=CFG['PC_JOIN_MIN_REL'])
                        recs = pc_join.build_records([_p['pc'][i] for i in _ix], [_p['pc_keys'][i] for i in _ix], chem, fragmod,
                                                     P.bits, fz=[_p['pc_fz'][i] for i in _ix])
                        recs = [r for r in recs if abs(r['mass'] - target) <= target * 30e-6]   # never the 'CCO' fallback
                except Exception as e:                                # scoring rerun: never lose the molecule
                    if not IS_RERUN:
                        raise
                    print('PC JOIN FAILED -> no injection', mid, repr(e)); recs = []; n_join_err += 1
                pc_join.set_injection(E, recs)
            try:
                C, X, F, names, info = V.run(spectra, target)
            except Exception as e:
                if not recs or not IS_RERUN:
                    raise
                print('JOINED RUN FAILED -> run without injection', mid, repr(e)); n_join_err += 1
                pc_join.set_injection(E, []); C, X, F, names, info = V.run(spectra, target)
            lib_max = float(info.get('lib_max', 0.0))
            if CFG['PC_JOIN_N'] > 0:
                inj = pc_join.injected_mask(E, C); _ji = pc_join.injected_info(E)
                n_join += len(_ji); n_join_mol += int(len(_ji) > 0)
                _out = {len(C.pid) - len(_ji) + t: d['pc_smiles'] for t, d in enumerate(_ji)}   # tier SMILES: one forward-model job per structure
                if (gi + 1) % 50 == 0:                                # per-SMILES caches of pid < 0 rows (33 MB per molecule at N 50)
                    for _fm, _c in (('fragnet', '_GEN'), ('frag2', '_GF'), ('frag2', '_GC')):
                        if _fm in V.mods and hasattr(V.mods[_fm], _c):
                            getattr(V.mods[_fm], _c).clear()
            if len(C.pid):
                score = rank_score(np.concatenate([X, F], 1), list(FEATURES) + list(names))
                if CFG['RANKER'] != 'public':
                    _a = (CFG['RANKER'], OUR_RK, C.formula, X, F, list(FEATURES), list(names), lib_max, rank_score)
                    if not IS_RERUN:      # commit runs: exercise the ungated branch on every molecule (smoke molecules
                        _dry = pc_join.choose_score(*_a, None, inj)       # are library hits); a bug must fail the commit
                        n_rk_diff += int(np.argmax(_dry) != np.argmax(score))
                    try:
                        score = pc_join.choose_score(*_a, CFG['RANKER_LIB_GATE'], inj)   # == ranker_infer.choose_score without injected rows
                    except Exception as e:                    # scoring rerun: never lose the molecule
                        if not IS_RERUN:
                            raise
                        print('RANKER FAILED -> public ranker', mid, repr(e)); n_rk_err += 1
                order = np.argsort(-score, kind='stable')
                seen = set()
                for i in order:
                    k = C.key[i]
                    if k in seen: continue
                    seen.add(k); smis.append(_out.get(int(i), C.smiles[i])); keys.append(k); scs.append(float(score[i])); forms.append(C.formula[i]); pids.append(int(C.pid[i]))
                    if len(smis) >= CFG['TOPN']: break
        except Exception as e:
            print('ERROR', mid, repr(e)); n_err += 1
    BASE[mid] = [smis, keys, lib_max, scs, forms, pids]
    if (gi + 1) % 25 == 0 or gi == len(mols) - 1:
        log(f'  {gi+1}/{len(mols)} molecules')
log('base lists done, errors', n_err, '| RANKER', CFG['RANKER'], 'gate', CFG['RANKER_LIB_GATE'], 'fallbacks', n_rk_err,
    '| ungated top-1 differs from the public ranker (commit runs):', n_rk_diff,
    '| PC join: injected rows', n_join, 'in', n_join_mol, 'molecules, failures', n_join_err)
assert IS_RERUN or (CFG['RANKER'] == 'public' and CFG['PC_JOIN_N'] == 0) or n_err == 0, \
    'base-list errors with RANKER != public or PC_JOIN_N > 0: fix before submitting'
dump('base', BASE)
```

## 9. Forward-model scores
ICEBERG and GLACIER predict a spectrum for each candidate; the score is the similarity to the measured spectra. Only
same-formula groups are scored, and only [M+H]+ and [M+Na]+ spectra are covered, under a time budget. Scores are only
computed here; the re-ranking is in the next cell. Any failure leaves the ranker order untouched.


```python
# ICEBERG + GLACIER scoring of same-formula candidate groups (base top-N ∪ engine-2 top-K ∪ gated PubChem lists).
# Scores only; the re-ranking itself happens in fusion_core. Any failure -> empty scores -> ranker order kept.
import gc
from rdkit import Chem as _Chem
from rdkit.Chem.rdMolDescriptors import CalcMolFormula as _CMF

def formula_of(s):
    try:
        return _CMF(_Chem.MolFromSmiles(s))
    except Exception:
        return s

ICE_SCORES, GL_SCORES, items, items_native, ice_fuse, gl_fuse = {}, {}, None, None, None, None
TOPN, LIB_TAU = CFG['TOPN'], CFG['LIB_TAU']
if CFG['USE_ICE']:
    try:
        ICE_PKG = os.path.dirname(find('casmi26-iceberg/**/ice_runner.py'))
        sys.path.append(ICE_PKG)
        import fuse as ice_fuse
        del V, E, bank; gc.collect(); torch.cuda.empty_cache()
        mids = sorted(BASE, key=lambda m: BASE[m][2])        # molecules without a library match first
        cands = {m: ice_fuse.ice_candidates(BASE[m][0], BASE[m][4], TOPN) for m in mids}
        if ENG and CFG['ICE_UNION']:
            K, n_add = CFG['ICE_UNION_K'], 0
            for m in mids:
                e = ENG.get(str(m))
                if not e or not e.get('smiles'):
                    continue
                bs, bk, bf = list(BASE[m][0][:TOPN]), set(BASE[m][1][:TOPN]), list(BASE[m][4][:TOPN])
                es = [s for s, k in zip(e['smiles'][:K], e['keys'][:K]) if k not in bk]
                extra = ice_fuse.ice_candidates(bs + es, bf + [formula_of(s) for s in es], len(bs) + len(es))
                before = len(cands.get(m, []))
                cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + extra))
                n_add += len(cands[m]) - before
            log('ICE_UNION: engine-2 candidates added:', n_add)
        if CFG['ICE_PC']:
            for m in mids:
                p = PC.get(m) if isinstance(PC, dict) else None
                if p and p.get('pc') and BASE[m][2] < LIB_TAU:
                    p['pc_form'] = [formula_of(x) for x in p['pc']]
                    extra = ice_fuse.ice_candidates(p['pc'], p['pc_form'], 25)
                    cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + extra))
        for m in mids:                           # VAL_EXTRAS: score the popularity-channel proposals as well
            p = PC_ALT.get(m) if isinstance(PC_ALT, dict) else None
            if p and p.get('pc') and BASE[m][2] < LIB_TAU:
                extra = ice_fuse.ice_candidates(p['pc'], [formula_of(x) for x in p['pc']], 25)
                cands[m] = list(dict.fromkeys(list(cands.get(m, [])) + extra))
        cands = {m: c for m, c in cands.items() if c}
        items = items_native = ice_fuse.build_ice_input(te, cands)
        if CFG['ICE_SURROGATE']:                 # [M+NH4]+ / water-loss spectra scored against the [M+H]+ prediction
            import surrogate
            items, _n = surrogate.extend_items(items_native, ice_fuse.build_ice_input(surrogate.rewrite(te), cands),
                                               CFG['ICE_SURROGATE'])
            log('ICE_SURROGATE', CFG['ICE_SURROGATE'], ': molecules with surrogate spectra', _n)
        log('ICE input molecules', len(items), 'candidates', sum(len(it['cands']) for it in items))
        ICE_SCORES = ice_fuse.run_ice(ICE_PKG, items, workdir=os.path.join(WORK, 'ice_work'), device='cuda',
                                      budget_s=ICE_BUDGET, site=os.path.join(WORK, 'ice_site'))
        meta_f = os.path.join(WORK, 'ice_work', 'ice_out.json.meta.json')
        if os.path.exists(meta_f):
            print('ICE meta', open(meta_f).read()[:1500])
    except Exception as e:
        ICE_SCORES = {}
        print('ICE FAILED -> ranker order kept:', repr(e))
if CFG['USE_GL'] and items:
    try:
        GL_PKG = os.path.dirname(find('casmi26-glacier/**/gl_runner.py'))
        sys.path.append(GL_PKG)
        import gl_fuse
        gc.collect(); torch.cuda.empty_cache()
        gl_items = items if CFG['GL_SURROGATE'] else items_native    # surrogate spectra are untested with GLACIER
        if CFG['GL_MH_ONLY']:                # GLACIER on [M+H]+ spectra only (wangpenghua v29: +0.006 on the LB)
            gl_items = [dict(it, spectra=[s for s in it['spectra'] if s.get('adduct') == '[M+H]+']) for it in items]
            gl_items = [it for it in gl_items if it['spectra']]
            log('GLACIER [M+H]+ only: molecules', len(gl_items), 'of', len(items))
        GL_SCORES = gl_fuse.run_gl(GL_PKG, gl_items, workdir=os.path.join(WORK, 'gl_work'), device='cuda',
                                   budget_s=CFG['GL_BUDGET'], site=os.path.join(WORK, 'ice_site'),
                                   wheels=os.path.join(ICE_PKG, 'wheels'))
        meta_f = os.path.join(WORK, 'gl_work', 'gl_out.json.meta.json')
        print('GL meta', open(meta_f).read()[:1500] if os.path.exists(meta_f) else 'missing')
    except Exception as e:
        GL_SCORES = {}
        print('GL FAILED -> ICE only:', repr(e))
FRAG = {}                                   # MetFrag-parsimony term (frag_rescore.py), CPU, ~1 ms per candidate
if CFG['FRAG_LAM'] > 0:
    try:
        import frag_rescore
        _t0 = time.time()
        for mid, _sub in te.groupby('molecule_id', sort=False):
            if mid not in BASE:
                continue
            _spec = [dict(mz=np.asarray(r.ms2_mzs, np.float64), it=np.asarray(r.ms2_normalized_intensities, np.float64),
                          adduct=r.adduct, prec=float(r.precursor_mz), mode=1 if r.ionization_mode == 'positive' else -1)
                     for r in _sub.itertuples()][:16]
            _c = list(dict.fromkeys(list(BASE[mid][0]) + list((ENG.get(str(mid)) or {}).get('smiles', [])) +
                                    list((PC.get(mid) or {}).get('pc', []))))
            FRAG[mid] = dict(zip(_c, frag_rescore.score_molecule(_spec, _c)))
        log('frag scores', len(FRAG), 'molecules', f'{time.time()-_t0:.0f}s')
    except Exception as e:
        FRAG = {}
        print('FRAG FAILED -> off:', repr(e))
GL_MH = {}                                    # VAL_EXTRAS: GLACIER on [M+H]+ spectra only, for local replay
if CFG['VALIDATION'] and CFG['VAL_EXTRAS'] and CFG['USE_GL'] and items and not CFG['GL_MH_ONLY'] and not IS_RERUN:
    try:
        _mh = [dict(it, spectra=[s for s in it['spectra'] if s.get('adduct') == '[M+H]+']) for it in items_native]
        _mh = [it for it in _mh if it['spectra']]
        gc.collect(); torch.cuda.empty_cache()
        GL_MH = gl_fuse.run_gl(GL_PKG, _mh, workdir=os.path.join(WORK, 'gl_work_mh'), device='cuda',
                               budget_s=CFG['GL_BUDGET'], site=os.path.join(WORK, 'ice_site'),
                               wheels=os.path.join(ICE_PKG, 'wheels'))
    except Exception as e:
        GL_MH = {}
        print('GL [M+H]+ EXTRA FAILED:', repr(e))
    log('GLACIER [M+H]+ only (extras): molecules', len(GL_MH))
    dump('gl_mh', GL_MH)
log('ICE molecules', len(ICE_SCORES), '| GL molecules', len(GL_SCORES), '| FRAG molecules', len(FRAG))
dump('ice', ICE_SCORES); dump('gl', GL_SCORES); dump('frag', FRAG); dump('pc', PC)   # pc again: now carries pc_form
```

## 10. Fusion and submission
`fusion_core.build_submission` does, per molecule:
1. **Stage A**: cut to `TOPN`, apply the popularity prior, then re-order inside each same-formula group with the
   forward-model scores (not for library hits), plus the fragment term where ICEBERG has no score.
2. **Stage B**: if the library does not know the compound, insert PubChem-only candidates at fixed slots, as in the
   reference.
3. **Stage C**: weighted reciprocal-rank fusion with engine 2 and one more forward-model re-rank of the fused list.

The result is at most 25 SMILES per molecule, de-duplicated by structure key, written to `submission.csv`.


```python
# fusion_core: ICE/GL re-rank -> gated PubChem merge -> engine-2 fusion -> submission.csv
samp = pd.read_csv(os.path.join(COMP, 'sample_submission.csv'))
MOL_ORDER = list(te.molecule_id.unique())
dump('meta', dict(mol_order=MOL_ORDER, sample_ids=list(samp.molecule_id), is_rerun=IS_RERUN, cfg=CFG))
POOL_POP = None
if CFG['POP_MU'] > 0:
    try:                                  # log1p(PubChem SIDs) + log1p(PubMed ids) per v2-pool row (pid order)
        _pp = os.path.dirname(find('casmi26-pubchem-popularity-prior/**/pool_lsid.npy'))
        POOL_POP = np.load(os.path.join(_pp, 'pool_lsid.npy')).astype(np.float64) + \
            np.load(os.path.join(_pp, 'pool_lpmid.npy')).astype(np.float64)
        log('popularity prior loaded', _pp, POOL_POP.shape)
    except Exception as e:
        print('POPULARITY PRIOR UNAVAILABLE -> off:', repr(e))
sub, FSTATS = fusion_core.build_submission(
    BASE, PC, ENG, ICE_SCORES, GL_SCORES, CFG, MOL_ORDER, list(samp.molecule_id),
    ice_fuse=ice_fuse, gl_fuse=gl_fuse, score_key=chem.score_key, formula=formula_of, log=print, pool_pop=POOL_POP,
    frag_scores=FRAG)
fusion_core.validate(sub, list(samp.molecule_id))
sub.to_csv(os.path.join(WORK, 'submission.csv'), index=False)
json.dump(dict(version=CFG['VERSION'], secs=round(time.time() - T0), stats=FSTATS, n_errors=n_err, families=MAN['families'],
               gpu=torch.cuda.get_device_name(0) if dev == 'cuda' else None, cfg=CFG),
          open(os.path.join(WORK, 'run_manifest.json'), 'w'), indent=1)
log('fusion stats', FSTATS)
log('done', sub.shape)
if os.path.exists(os.path.join(STAGE, 'val_labels.csv')):   # validation mode: score with the metric's key
    from rdkit.Chem.MolStandardize import rdMolStandardize
    _taut = rdMolStandardize.TautomerEnumerator()
    def _ik14(s):
        try:
            return _Chem.MolToInchiKey(_taut.Canonicalize(_Chem.MolFromSmiles(s)))[:14]
        except Exception:
            return None
    _lab = pd.read_csv(os.path.join(STAGE, 'val_labels.csv')).set_index('molecule_id')
    _rr = []
    for _m, _s in zip(sub.molecule_id, sub.smiles):
        _t = _ik14(_lab.smiles[_m]); _r = 0.0
        for _i, _g in enumerate(_s.split(';'), 1):
            if _ik14(_g) == _t:
                _r = 1.0 / _i; break
        _rr.append(_r)
    log(f'VALIDATION MRR@25 = {np.mean(_rr):.4f} | top1 {np.mean(np.array(_rr) == 1):.3f} | hit@25 {np.mean(np.array(_rr) > 0):.3f} | n {len(_rr)}')
# the ICE/GLACIER wheel site holds thousands of files: drop it so the version's output stays small to list/download
import shutil; shutil.rmtree(os.path.join(WORK, 'ice_site'), ignore_errors=True)
sub.head()
```
